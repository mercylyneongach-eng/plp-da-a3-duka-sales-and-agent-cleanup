**Analyst Role:** Junior Data Analyst

### Project Objective

The objective of this analysis is to explore Duka Electronics' sales data, evaluate branch and product performance, compare sales patterns, identify revenue trends and provide actionable business insights to support management decision-making.

```python
import pandas as pd
import numpy as np

# Set random seed
rng = np.random.default_rng(2026)

# Number of orders
n = 1000

# Product prices
products = {
    "Phone": 15000,
    "Laptop": 65000,
    "Headphones": 2500,
    "Tablet": 28000,
    "Charger": 1200
}

# Generate sales dataset
sales = pd.DataFrame({
    "order_id": range(1001, 1001 + n),

    "order_date": (
        pd.to_datetime("2026-01-01")
        + pd.to_timedelta(
            rng.integers(0, 180, n), unit="D"
        )
    ),

    "branch": rng.choice(
        ["Nairobi", "Mombasa", "Kisumu", "Nakuru", "Eldoret"],
        n,
        p=[0.4, 0.2, 0.15, 0.15, 0.1]
    ),

    "product": rng.choice(list(products), n),

    "quantity": rng.integers(1, 6, n),

    "channel": rng.choice(
        ["In-store", "Online", "Phone order"],
        n,
        p=[0.55, 0.35, 0.10]
    )
})

# Add unit prices
sales["unit_price"] = sales["product"].map(products)

# Calculate revenue
sales["revenue"] = sales["quantity"] * sales["unit_price"]

# Create derived columns
sales["month"] = sales["order_date"].dt.month_name()
sales["weekday"] = sales["order_date"].dt.day_name()
sales["is_weekend"] = sales["order_date"].dt.dayofweek >= 5

sales["price_band"] = np.where(
    sales["unit_price"] >= 28000,
    "Premium",
    "Value"
)

# Save dataset
sales.to_csv("duka_sales.csv", index=False)

print("Duka Electronics dataset regenerated successfully!")
print("Dataset shape:", sales.shape)

sales.head()
```

```python
# 1. Dataset dimensions
print("Dataset shape:", sales.shape)

# 2. Column data types
print("\nData types:")
print(sales.dtypes)

# 3. Missing values
print("\nMissing values:")
print(sales.isnull().sum())

# 4. Date range
print("\nDate range:")
print("Start:", sales["order_date"].min())
print("End:", sales["order_date"].max())

# 5. Key categories
print("\nBranches:")
print(sales["branch"].unique())

print("\nProducts:")
print(sales["product"].unique())

print("\nSales channels:")
print(sales["channel"].unique())
```

## Task 3: Branch Performance Analysis

This section evaluates the performance of the five Duka Electronics branches by comparing their total revenue and number of orders. The objective is to identify the strongest-performing branches and determine whether revenue rankings correspond with order-volume rankings.

```python
# Rank branches by total revenue
branch_revenue = (
    sales.groupby("branch")["revenue"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

branch_revenue.columns = ["Branch", "Total Revenue (KES)"]

print("BRANCH RANKING BY REVENUE")
display(branch_revenue)


# Rank branches by number of orders
branch_orders = (
    sales.groupby("branch")["order_id"]
    .count()
    .sort_values(ascending=False)
    .reset_index()
)

branch_orders.columns = ["Branch", "Number of Orders"]

print("BRANCH RANKING BY NUMBER OF ORDERS")
display(branch_orders)
```

### Interpretation of Branch Performance

Nairobi branch recorded the highest performance, generating KES 26,106,100 from 376 orders. Mombasa followed with KES 14,003,700 from 186 orders, while Nakuru generated KES 11,970,400 from 162 orders. Kisumu recorded KES 10,868,100 from 159 orders, and Eldoret had the lowest performance, generating KES 8,844,200 from 117 orders.

The rankings by total revenue and number of orders are identical. This indicates that branches with higher order volumes also generated higher total revenue during the reporting period. However, the revenue generated per order may differ between branches depending on the products purchased and quantities sold.

Nairobi's leading position suggests that it is the strongest sales branch in terms of both transaction volume and total revenue. Management should investigate the factors behind its performance and consider whether successful practices can be applied to other branches.

## Task 4: Product Performance Analysis

This section identifies the three best-performing products based on total revenue and total units sold. Comparing the two rankings helps determine whether the products generating the most income are also the products with the highest sales volumes.

```python
# Rank products by total revenue
product_revenue = (
    sales.groupby("product")
    .agg(
        Total_Revenue=("revenue", "sum"),
        Units_Sold=("quantity", "sum")
    )
    .sort_values("Total_Revenue", ascending=False)
)

print("TOP THREE PRODUCTS BY REVENUE")
display(product_revenue.head(3))


# Rank products by units sold
product_units = (
    sales.groupby("product")["quantity"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

product_units.columns = ["Product", "Units Sold"]

print("TOP THREE PRODUCTS BY UNITS SOLD")
display(product_units.head(3))
```

### Interpretation of Product Performance

Laptops generated the highest revenue at KES 41,535,000, despite selling 639 units. Tablets ranked second in revenue, generating KES 19,348,000, but recorded the highest sales volume at 691 units. Phones ranked third in both revenue and units sold, with KES 8,880,000 generated from 592 units.

The difference between the revenue and unit-sales rankings is mainly explained by differences in product prices. Laptops have a higher unit price of KES 65,000, allowing them to generate more revenue even when fewer units are sold than tablets. Tablets, priced at KES 28,000, recorded higher unit sales but generated less total revenue.

This suggests that laptops are the strongest contributors to revenue, while tablets are the most popular product by quantity sold. Management should maintain laptop availability to protect revenue performance while ensuring sufficient tablet stock to meet customer demand.

## Task 5: Weekend and Weekday Sales Comparison

This section compares sales performance between weekdays and weekends using order volume, total revenue and average revenue per order. The analysis will help management understand customer purchasing patterns and identify periods of stronger sales activity.

```python
# Group sales by weekend and weekday
weekend_comparison = (
    sales.groupby("is_weekend")
    .agg(
        Number_of_Orders=("order_id", "count"),
        Total_Revenue=("revenue", "sum"),
        Average_Order_Revenue=("revenue", "mean")
    )
    .reset_index()
)

# Replace Boolean values with readable labels
weekend_comparison["is_weekend"] = weekend_comparison[
    "is_weekend"
].map({
    True: "Weekend",
    False: "Weekday"
})

display(weekend_comparison)
```

## Task 6: Monthly Revenue Performance

This task analyses monthly sales revenue for Duka Electronics between January and June 2026. The objective is to identify the best-performing and worst-performing months based on total revenue and calculate the percentage difference between them.

```python
monthly_revenue = (
    sales.groupby("month")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

monthly_revenue
```

```python
best_month = monthly_revenue.idxmax()
worst_month = monthly_revenue.idxmin()

best_revenue = monthly_revenue.max()
worst_revenue = monthly_revenue.min()

percentage_difference = (
    (best_revenue - worst_revenue) / worst_revenue
) * 100

print("BEST AND WORST MONTHLY PERFORMANCE")
print("-" * 40)

print(f"Best Month: {best_month}")
print(f"Highest Revenue: KES {best_revenue:,.2f}")

print(f"\nWorst Month: {worst_month}")
print(f"Lowest Revenue: KES {worst_revenue:,.2f}")

print(f"\nPercentage Difference: {percentage_difference:.2f}%")
```

### Interpretation

April was the best-performing month, generating the highest revenue of KES 13,518,300, while June was the worst-performing month, recording KES 10,903,400. The revenue generated in April was 23.98% higher than the revenue generated in June. This indicates a decline in sales revenue towards the end of the six-month period. Duka Electronics should investigate the factors contributing to the lower June performance and consider strategies such as targeted promotions, improved stock management and customer engagement to strengthen sales.

### Task 7: Premium Online Orders from Nairobi and Mombasa

This task identifies Premium orders made through the Online sales channel at the Nairobi and Mombasa branches. The analysis determines the number of qualifying orders and their total revenue. The filtered records are also exported to a CSV file for further use and reporting.

```python
premium_online_coast_capital = sales[
    (sales["price_band"] == "Premium") &
    (sales["channel"] == "Online") &
    (sales["branch"].isin(["Nairobi", "Mombasa"]))
].copy()

premium_online_coast_capital.head()
```

```python
number_of_orders = len(premium_online_coast_capital)

total_revenue = premium_online_coast_capital["revenue"].sum()

print("PREMIUM ONLINE ORDERS: NAIROBI AND MOMBASA")
print("-" * 50)

print(f"Number of Orders: {number_of_orders:,}")
print(f"Total Revenue: KES {total_revenue:,.2f}")
```

### Interpretation

The analysis identified 83 Premium orders made through the Online sales channel at the Nairobi and Mombasa branches, generating a total revenue of KES 10,218,000. These orders represent a significant contribution to sales revenue from high-value products sold through online transactions. The findings highlight the importance of the online sales channel in generating revenue from Premium products in the two major branches. Duka Electronics should continue strengthening its online sales strategy and ensure adequate availability of high-value products to meet customer demand.
